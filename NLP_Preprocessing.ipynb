```python
import pandas as pd
import re
from google.colab import drive

# Check if bnltk is installed, if not, install it
try:
    from bnltk.tokenize import Tokenizers
except ImportError:
    print("bnltk not found. Installing...")
    !pip install bnltk
    from bnltk.tokenize import Tokenizers

# Mount Google Drive (check if already mounted)
try:
    drive.mount('/content/drive', force_remount=True)
    print("Google Drive mounted successfully.")
except:
    print("Google Drive already mounted.")

# Load CSV file
file_path = '/content/drive/MyDrive/Data/merged_y_f_comments/fb_comments_ts.csv'
df = pd.read_csv(file_path)

# Convert 'Date & Time' to datetime
df['Date & Time'] = pd.to_datetime(df['Date & Time'], errors='coerce')

# Drop rows with invalid or missing dates
df.dropna(subset=['Date & Time'], inplace=True)

# Sort data by 'Date & Time'
df.sort_values('Date & Time', inplace=True)

# Initialize Tokenizer
tokenizer = Tokenizers()

# Bangla Stopwords
bangla_stopwords = ['এই', 'ওই', 'এবং', 'কিন্তু', 'তবে', 'যে', 'তা', 'তুমি', 'আমি']

# Cleaning function
def clean_bangla_text(text):
    if pd.isnull(text):
        return ""

    text = re.sub(r'http\S+|www\S+', '', text)   # Remove URLs
    text = re.sub(r'[^\u0980-\u09FF\s]', '', text)   # Remove special characters
    text = re.sub(r'[০-৯0-9]', '', text)   # Remove digits
    text = re.sub(r'\b[a-zA-Z]+\b', '', text)   # Remove English words
    text = re.sub(r'[়]+', '', text)   # Normalize Bangla text

    # Tokenization and stopword removal
    tokens = tokenizer.bn_word_tokenizer(text)
    tokens = [word for word in tokens if word not in bangla_stopwords]

    return ' '.join(tokens)

# Apply cleaning
df['cleaned_comments'] = df['Comment'].apply(clean_bangla_text)

# Remove empty cleaned comments
df = df[df['cleaned_comments'].str.strip() != ""]

# Save cleaned data
df.to_csv('/content/drive/MyDrive/Data/Final_Merged/cleaned_ts_comments.csv', index=False)

print("Time series cleaning completed. File saved as 'cleaned_ts_comments.csv'.")
```

```python
from collections import Counter

# Combine all comments into one string
all_comments = ' '.join(df['cleaned_comments'].dropna())

# Split into words
all_words = all_comments.split()

# 1. Total number of words
total_words = len(all_words)
print(f"Total number of words in all comments: {total_words}")

# 2. Most common words
common_words = Counter(all_words).most_common(20)  # Top 20 most common words
print("Top 20 most common words:")
for word, count in common_words:
    print(f"{word}: {count}")
```
