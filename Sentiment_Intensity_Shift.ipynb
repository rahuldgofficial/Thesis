```python
#Time Series Forecasting
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.metrics import mean_absolute_error
from statsmodels.tsa.statespace.sarimax import SARIMAX
from prophet import Prophet
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Conv1D, Flatten
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import MeanAbsoluteError
from google.colab import drive


# 1) Load data
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/comments_new.csv',
                 parse_dates=['date_time']).set_index('date_time')
daily = df['sentiment_intensity'].resample('D').mean().fillna(method='ffill')

# Train/test split
split = int(0.8 * len(daily))
train, test = daily.iloc[:split], daily.iloc[split:]
test_dates = test.index
look_back = 7

# 2) Sequence helper
def create_sequences(data, look_back):
    X, y = [], []
    for i in range(look_back, len(data)):
        X.append(data[i-look_back:i])
        y.append(data[i])
    return np.array(X), np.array(y)

# 3) SARIMA
sarima_model = SARIMAX(train, order=(1,1,1), seasonal_order=(1,0,1,7)).fit(disp=False)
sarima_fcst = sarima_model.get_forecast(steps=len(test))
sarima_pred = sarima_fcst.predicted_mean  # Use this name consistently
sarima_ci   = sarima_fcst.conf_int(alpha=0.05)

# 4) Prophet
df_prophet = train.reset_index().rename(columns={'date_time':'ds','sentiment_intensity':'y'})
prophet_model = Prophet(daily_seasonality=False, weekly_seasonality=True, interval_width=0.95)
prophet_model.fit(df_prophet)
future = prophet_model.make_future_dataframe(periods=len(test), freq='D')
prophet_forecast = prophet_model.predict(future).set_index('ds')

# 5) LSTM
scaler_lstm = MinMaxScaler()
train_scaled = scaler_lstm.fit_transform(train.values.reshape(-1, 1)).flatten()
X_lstm, y_lstm = create_sequences(train_scaled, look_back)
X_lstm = X_lstm.reshape(-1, look_back, 1)

lstm_model = Sequential([
    LSTM(50, activation='relu', input_shape=(look_back, 1)),
    Dense(1)
])
lstm_model.compile(optimizer=Adam(0.001), loss=MeanAbsoluteError())
lstm_model.fit(X_lstm, y_lstm, epochs=50, batch_size=32, verbose=0)

# Forecast with LSTM
test_scaled = scaler_lstm.transform(test.values.reshape(-1, 1)).flatten()
X_test_lstm, _ = create_sequences(
    np.concatenate([train_scaled[-look_back:], test_scaled]),
    look_back
)
lstm_preds = lstm_model.predict(X_test_lstm).flatten()
lstm_preds = scaler_lstm.inverse_transform(lstm_preds.reshape(-1,1)).flatten()

# 6) TCN
scaler_tcn = MinMaxScaler()
train_scaled_tcn = scaler_tcn.fit_transform(train.values.reshape(-1, 1)).flatten()
X_tcn, y_tcn = create_sequences(train_scaled_tcn, look_back)
X_tcn = X_tcn.reshape(-1, look_back, 1)

tcn_model = Sequential([
    Conv1D(64, 2, activation='relu', padding='causal', input_shape=(look_back, 1)),
    Flatten(),
    Dense(50, activation='relu'),
    Dense(1)
])
tcn_model.compile(optimizer=Adam(0.001), loss=MeanAbsoluteError())
tcn_model.fit(X_tcn, y_tcn, epochs=50, batch_size=32, verbose=0)

# Forecast with TCN
test_scaled_tcn = scaler_tcn.transform(test.values.reshape(-1, 1)).flatten()
X_test_tcn, _ = create_sequences(
    np.concatenate([train_scaled_tcn[-look_back:], test_scaled_tcn]),
    look_back
)
tcn_preds = tcn_model.predict(X_test_tcn).flatten()
tcn_preds = scaler_tcn.inverse_transform(tcn_preds.reshape(-1,1)).flatten()

# 7) Cross-Validation Errors
sarima_errors, prophet_errors, lstm_errors, tcn_errors = [], [], [], []
for i in range(split, len(daily)-1):
    train_cv = daily.iloc[:i]
    true_val = daily.iloc[i]

    # SARIMA CV
    m_sarima = SARIMAX(train_cv, order=(1,1,1), seasonal_order=(1,0,1,7)).fit(disp=False)
    e_sarima = m_sarima.get_forecast(steps=1).predicted_mean.iloc[0]
    sarima_errors.append(abs(true_val - e_sarima))

    # Prophet CV
    df_cv = train_cv.reset_index().rename(columns={'date_time':'ds','sentiment_intensity':'y'})
    m_prophet = Prophet(daily_seasonality=False, weekly_seasonality=True).fit(df_cv)
    f_cv = m_prophet.make_future_dataframe(periods=1, freq='D')
    p_cv = m_prophet.predict(f_cv)['yhat'].iloc[-1]
    prophet_errors.append(abs(true_val - p_cv))

    # LSTM CV
    scaler_cv = MinMaxScaler()
    s_cv = scaler_cv.fit_transform(train_cv.values.reshape(-1,1)).flatten()
    if len(s_cv) >= look_back:
        X_cv, _ = create_sequences(s_cv, look_back)
        X_cv = X_cv.reshape(-1, look_back, 1)
        m_lstm = Sequential([LSTM(50), Dense(1)])
        m_lstm.compile(optimizer=Adam(0.001), loss=MeanAbsoluteError())
        m_lstm.fit(X_cv, s_cv[look_back:], epochs=50, batch_size=32, verbose=0)
        f_lstm = scaler_cv.inverse_transform(
            m_lstm.predict(X_cv[-1:].reshape(1, look_back, 1))
        )[0,0]
        lstm_errors.append(abs(true_val - f_lstm))

    # TCN CV
    m_tcn = Sequential([
        Conv1D(64, 2, activation='relu', padding='causal', input_shape=(look_back,1)),
        Flatten(), Dense(1)
    ])
    m_tcn.compile(optimizer=Adam(0.001), loss=MeanAbsoluteError())
    m_tcn.fit(X_cv, s_cv[look_back:], epochs=50, batch_size=32, verbose=0)
    f_tcn = scaler_cv.inverse_transform(
        m_tcn.predict(X_cv[-1:].reshape(1, look_back, 1))
    )[0,0]
    tcn_errors.append(abs(true_val - f_tcn))

# 8) Enhanced Visualization
plt.figure(figsize=(14,8))
plt.plot(train.index, train, color='gray', alpha=0.5, label='Train')
plt.plot(test_dates, test,    lw=2,               label='Actual')

# SARIMA
plt.plot(test_dates, sarima_pred,    '--', color='orange', label='SARIMA')
plt.fill_between(test_dates,
                 sarima_ci.iloc[:,0], sarima_ci.iloc[:,1],
                 color='orange', alpha=0.2)

# Prophet
plt.plot(test_dates, prophet_forecast.loc[test_dates,'yhat'],
         '--', color='green', label='Prophet')
plt.fill_between(test_dates,
                 prophet_forecast.loc[test_dates,'yhat_lower'],
                 prophet_forecast.loc[test_dates,'yhat_upper'],
                 color='green', alpha=0.2)

# LSTM & TCN
plt.plot(test_dates[:len(lstm_preds)], lstm_preds,
         '-o', color='blue', label='LSTM')
plt.plot(test_dates[:len(tcn_preds)], tcn_preds,
         '-s', color='red', label='TCN')

plt.title("Daily Sentiment Intensity: Forecasts & CV MAEs")
plt.xlabel("Date"); plt.ylabel("Sentiment Intensity")
plt.xticks(rotation=45); plt.tight_layout()
plt.legend(loc='upper left', bbox_to_anchor=(1,1))
plt.show()

# 9) Print CV Results
print("Cross-Validation MAE:")
print(f"SARIMA:  {np.nanmean(sarima_errors):.3f}")
print(f"Prophet: {np.nanmean(prophet_errors):.3f}")
print(f"LSTM:    {np.nanmean(lstm_errors):.3f}")
print(f"TCN:     {np.nanmean(tcn_errors):.3f}")
```
